```python
import os

# Change this line to include all of the AFQM characters.
char_list = ['eureka','knockt','ninja','rend', 'sarissae', 'sarissae2']

# Change this line to wherever the AFQM scripts are.
home = 'C:/Users/EllieVenn/Documents/AFQM 2'
```

```python
# This grabs the movelist for each character so that their scripts can be hunted later.

move_list = list()
for h in os.listdir(home):
    if('character_init_' in h):
        with open(os.path.join(home, h), 'r') as f:
            for line in f:
                r = line
                if('my_attacks' in r and '-1' not in r):
                    print(r)
                    move_list.append(r.split(', ')[-1].strip(');\n'))
```

```python
# This creates a dict of all the attacks, split by character and move.

move_data = dict()
for char in char_list:
    move_data[char.title()] = dict()
for m in move_list:
    with open(os.path.join(home, 'gml_Globalscript_' + m + '.gml'), 'r') as f:
        s = m.rsplit('_', maxsplit=1)
        chara = s[1].title()
        move = s[0].title()
        if(move == 'Dash_Attack'):
            move = 'Dattack'
        if(move == 'Getup_Attack'):
            move = 'Gattack'
        if(move == 'Final'):
            move = 'Burst'
        move_data[chara][move] = list()
        for line in f:
            r = line
            if 'hitbox_create_melee(' in r:
                props_list = r.split('(')[-1].strip(');\n').split(',')
                obj = {
                    'dmg': props_list[4],
                    'bkb': props_list[5],
                    'kbs': props_list[6],
                    'hitlag': props_list[7],
                    'angle': props_list[8],
                    'active_frame_count': props_list[9]
                }
                if(obj not in move_data[chara][move]):
                    move_data[chara][move].append(obj)
            if 'hitbox_create_projectile(' in r:
                props_list = r.split('(')[-1].strip(');\n').split(',')
                obj = {
                    'dmg': props_list[4],
                    'bkb': props_list[5],
                    'kbs': props_list[6],
                    'angle': props_list[7],
                    'active_frame_count': props_list[8],
                    'projectile': True
                }
                if(obj not in move_data[chara][move]):
                    move_data[chara][move].append(obj)
            if 'hitbox_create_projectile_custom(' in r:
                props_list = r.split('(')[-1].strip(');\n').split(',')
                obj = {
                    'dmg': props_list[5],
                    'bkb': props_list[6],
                    'kbs': props_list[7],
                    'angle': props_list[8],
                    'active_frame_count': props_list[9],
                    'projectile': True
                }
                if(obj not in move_data[chara][move]):
                    move_data[chara][move].append(obj)
```

```python
# This prints the hitbox information in wiki-friendly format.

for chara, move in move_data.items():
    for move_id, hits in move.items():
        index = 0
        for hit in hits:
            print(f'''{{{{AFQM-HitData
|{'chara' :<10} = {chara}
|{'attack' :<10} = {move_id}
|{'hit_id' :<10} = {index}
|{'dmg' :<10} = {hit['dmg'].strip()}
|{'bkb' :<10} = {hit['bkb'].strip()}
|{'kbs' :<10} = {hit['kbs'].strip()}
|{'angle' :<10} = {hit['angle'].strip()}''')
            if('hitlag' in hit.keys()):
                print(f'|{'hitlag' :<10} = {hit['hitlag'].strip()}')
            if('projectile' in hit.keys()):
                print(f'|{'projectile' :<10} = True')
            print(f'}}}}')
            index += 1
```

```python
# This creates move mode placeholders for each move. Only necessary for new characters.

for chara, move in move_data.items():
    for move_id, hits in move.items():
        print(f'''{{{{AFQM-MoveMode
| chara = {chara}
| attack = {move_id}
| image = AFQM_{chara}_{move_id}_0.png
| totalActive =
| totalDuration = 
| hitID = {';'.join(map(lambda x: str(x), list(range(len(hits)))))}
| hitActive =
}}}}''')
```

```python
# This extracts all of the character properties.

char_props = dict()
for h in os.listdir(home):
    if('character_init_' in h):
        with open(os.path.join(home, h), 'r') as f:
            character_name = h.split('_')[-1].rstrip('.gml')
            char_props[character_name] = dict()
            flag = False
            for line in f:
                if('if (_set_properties)' in line):
                    flag = True
                elif(' = ' in line and flag):
                    char_props[character_name][line.split(' = ')[0].lstrip()] = line.split(' = ')[1].rstrip(';\n')
                elif('}' in line):
                    flag = False
```

```python
# This prints every character's info in a data-friendly format.

for k, v in char_props.items():
    output = f'''{{{{AFQM-Character
| {'chara': <30} = {k.title()}'''
    for k2, v2 in v.items():
        if(k2 in ['jostle_strength','waveland_time','collision_box','hit_cancel_speed','airdodge_speed','airdodge_startup','airdodge_active','airdodge_endlag', 'hurtbox_sprite', 'hurtbox_crouch_sprite', 'parry_press_script', 'draw_script', 'roll_speed','roll_startup','roll_active','roll_endlag','getup_active','getup_endlag','tech_active','tech_endlag','techroll_speed','techroll_startup','techroll_active','techroll_endlag','helpless_accel','helpless_max_speed','item_hold_x_default','item_hold_y_default','item_hold_r_default','weight_multiplier','ledge_jump_vsp','ledge_jump_hsp','ledge_jump_time','ledge_getup_time','ledge_getup_finish_x','ledge_getup_finish_y','ledge_roll_time','ledge_attack_time','ledge_hang_relative_x','ledge_hang_relative_y','parry_press_startup','parry_press_active','parry_press_endlag','parry_press_trigger_time','parry_press_script','can_wall_jump','wall_jump_startup','wall_jump_time','wall_jump_hsp','wall_jump_vsp','max_wall_jumps']):
            continue
        output += f'\n| {k2: <30} = {v2}'
    output += '\n}}\n'
    print(output)
```
